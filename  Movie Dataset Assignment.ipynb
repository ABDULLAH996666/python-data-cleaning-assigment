### **Question 23**

### **Question 36**

```python
data["Gross"] = pd.to_numeric(
    data["Gross"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)

drama_gross = data[
    data["Genre"].str.contains("Drama", case=False, na=False)
]["Gross"].dropna()
action_gross = data[
    data["Genre"].str.contains("Action", case=False, na=False)
]["Gross"].dropna()
plt.figure(figsize=(8, 6))
plt.boxplot(
    [drama_gross, action_gross],
    tick_labels=["Drama", "Action"]
)

plt.title("Gross Revenue Comparison: Drama vs Action")
plt.xlabel("Genre")
plt.ylabel("Gross Revenue")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = data.nlargest(5, "IMDB_Rating").copy()

top5["Meta_score"] = pd.to_numeric(
    top5["Meta_score"], errors="coerce"
)

top5["IMDB_Rating_Scaled"] = top5["IMDB_Rating"] * 10

top5 = top5.dropna(subset=["Meta_score"])

x = range(len(top5))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(
    [i - width / 2 for i in x],
    top5["Meta_score"],
    width,
    label="Meta Score"
)

plt.bar(
    [i + width / 2 for i in x],
    top5["IMDB_Rating_Scaled"],
    width,
    label="IMDb Rating (scaled to 100)"
)

plt.xticks(x, top5["Series_Title"], rotation=30, ha="right")
plt.ylabel("Score (out of 100)")
plt.title("Top 5 Highest-Rated Movies: Meta Score vs IMDb Rating")
plt.legend()
plt.tight_layout()
plt.show()
```
